```python
#join the title with the ticket description
data = data.loc[:, ~data.columns.duplicated(keep="first")]
print(data.info())
```

```python
print(data.info())
data['ticket about'] = data['Title'] + ' ' + data['hallo']
```

```python
#rename again the column to: Description
data.rename(columns={'ticket about': 'description','Your ticket is about:': 'about'}, inplace=True)
print(data.info())
print(data.head())
```

```python
print(data.info())
print(data.head())
X = data['description'] 
y = data['about']
```

Split your data already to be sure that no data likeage will happens

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state = 4)
```

```python
train = pd.concat([X_train, y_train], axis =1)
train.to_csv('train_tickets.csv')
```

```python
test = pd.concat([X_test, y_test], axis =1)
test.to_csv('test_tickets.csv')
```
